## Logic - Read Students and GPA, check whether on Dean's List or Honor Roll

the loop below represents this logic:

1. while True: -- repeats until `break` is reached.  True is endless loop.
2. student_last_name = input(...) -- reads and stores the student's last name.
3. if student_last_name == "zzz":` checks whether time to break loop
4. break -- ends the loop immediately, without further input, go to finished lines
5. student_first_name = input(...) -- reads and stores the student's first name
6. GPA = float(input(...)) -- reads the GPA and converts the entered text to a floating-point number
7. if GPA > 3.5: == checks whether the GPA is greater than 3.5. then "print(...)" -- displays the student's first name, a space, last name, and Dean's List message.
9. elif gpa > 3.25: -- checks whether the GPA is greater than 3.25 only when the Dean's List condition was false
10. The second `print(...) -- displays the student's first name, a space, last name, and Honor Role message.


```python
while True:
    student_last_name = input("Enter student's last name (zzz to quit): ")

    if student_last_name == "zzz":
        break

    student_first_name = input("Enter student's first name: ")
    GPA = float(input("Enter student's GPA: "))

    if GPA > 3.5:
        print(f"{student_first_name} {student_last_name} is on Dean's List")
    elif GPA > 3.25:
        print(f"{student_first_name} {student_last_name} is on Honor Role")
        
```
